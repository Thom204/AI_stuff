```python

import os
import pandas as pd
import requests
from tqdm import tqdm

def download_images(csv_file, download_folder):

    # Create folder
    os.makedirs(download_folder, exist_ok=True)

    # Read URLs directly, preserving commas
    with open(csv_file, 'r', encoding='utf-8-sig') as f:
        urls = [line.strip() for line in f if line.strip()]

    # Remove header
    if urls and urls[0].lower() == 'image url':
        urls = urls[1:]

    # Create DataFrame
    df = pd.DataFrame({'Image URL': urls})

    print("Columns detected:")
    print(df.columns.tolist())

    print("Total URLs:", len(df))

    display(df.head())

    # Download images
    for i, url in tqdm(
        enumerate(df['Image URL']),
        total=len(df),
        desc="Downloading images"
    ):

        try:
            filename = os.path.join(
                download_folder,
                f"image_{i}.jpg"
            )

            # Skip existing files
            if os.path.exists(filename):
                continue

            response = requests.get(
                url,
                stream=True,
                timeout=30
            )

            response.raise_for_status()

            # Verify that the response is an image
            content_type = response.headers.get(
                'Content-Type', ''
            ).lower()

            if 'image' not in content_type:
                print(f"Not an image: {url}")
                continue

            with open(filename, 'wb') as f:
                for chunk in response.iter_content(8192):
                    if chunk:
                        f.write(chunk)

        except Exception as e:
            print(f"Error downloading image {i}: {e}")

    print("\nDownload process completed!")


# Parameters
csv_file = 'nun_images.csv'
download_folder = 'nun_images'

# Execute
download_images(csv_file, download_folder)
```
